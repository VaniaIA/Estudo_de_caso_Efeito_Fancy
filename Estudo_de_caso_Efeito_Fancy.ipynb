```python
import pandas as pd
```

```python
df_vendas = pd.read_csv("/content/fancygeneration/vendas_pedidos.csv")
df_clientes = pd.read_csv("/content/fancygeneration/perfil_clientes.csv")
df_catalogo = pd.read_csv("/content/fancygeneration/catalogo_produtos.csv")
```

```python
df_vendas_clientes = pd.merge(df_vendas, df_clientes, left_on= "id_cliente", right_on = "id_cliente", how = "inner")
df_vendas_clientes_catalogo = pd.merge(df_vendas_clientes, df_catalogo, left_on= "sku_produto", right_on = "sku_produto", how = "inner")
df_vendas_clientes_catalogo["Lucro Bruto"] = (df_vendas_clientes_catalogo["preco_venda"] - df_vendas_clientes_catalogo["custo_producao"]) * df_vendas_clientes_catalogo["quantidade"]

display(df_vendas_clientes_catalogo)
```

```python
# Total de itens coluna quantidade e cálculo do preço de vendas

# Realiza soma de cada coluna mencionada

df_quantidade_total = df_vendas_clientes_catalogo["quantidade"].sum()
df_preco_venda_total = df_vendas_clientes_catalogo["preco_venda"].sum()

# Ticket médio para todas as vendas

df_ticket_medio = df_preco_venda_total / df_quantidade_total

print(f" O total de itens é de: {df_quantidade_total}")
print(f" O preço de venda total é de: {df_preco_venda_total:.2f}")
print(f" O ticket médio é de: {df_ticket_medio:.2f}")
```

```python
# Cálculo de cliente

df_total_de_clientes = df_vendas_clientes_catalogo["id_cliente"].nunique()
print(f" O total de clientes é de: {df_total_de_clientes}")
```

```python
# Ticket médio por Cliente

df_ticket_medio_por_cliente = df_preco_venda_total / df_total_de_clientes
print(f" O ticket médio por cliente é de: {df_ticket_medio_por_cliente:.2f}")
```

```python
df_vendas_clientes_catalogo.to_csv("vendas_clientes_catalogo.csv", index = False)
```
